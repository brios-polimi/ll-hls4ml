# Interactive hierarchical CDFG explorer

The single panel below selects and loads any `data/graphs/{kernel}/archive_{n}/*.json` graph, then renders a bounded neighborhood. Its levels, node types, and relations are derived from the loaded graph, so a control is only offered when that view can draw it.

Pragmas remain explicit nodes in node, block, and function views. Search for a pragma and select it as the center to inspect its real `applies_to` neighborhood. At hierarchy levels, optional `data_dependency` edges are derived only from real instruction → variable → instruction def-use paths; they are not stored as block/function data edges. Function names are demangled when `c++filt` is available, the selected center has a black outline, and the legend identifies visible colors.
