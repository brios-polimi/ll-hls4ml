# Interactive hierarchical CDFG explorer

The single panel below selects and loads any `data/graphs/{kernel}/archive_{n}/*.json` graph, then renders a bounded neighborhood. Its levels, node types, and relations are derived from the loaded graph, so a control is only offered when that view can draw it.

Hierarchy layout follows the structure of LLVM rather than applying one generic graph layout: function calls form a tree, each function owns a ranked block CFG, and node view arranges SSA producers, values, and consumers inside block regions. Shared high-degree constants sit on input rails and no longer expand an otherwise local neighborhood. Hover a node for its full LLVM context; click it for persistent details; drag it to resolve a local overlap; use the canvas toolbar to pan, zoom, and reset. Pragmas remain explicit at every level, and optional projected `data_dependency` edges are derived only from real instruction → variable → instruction paths.
